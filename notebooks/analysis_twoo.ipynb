```python
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------
# 1. LOAD DATA (you already have this)
# ----------------------------------------------------

# order_header_df = pd.read_csv('...')
# app_clicks = pd.read_csv('...')
# funnel_df = pd.read_csv('...')

# Ensure date columns parse
order_header_df['createdate'] = pd.to_datetime(order_header_df['createdate'], errors='ignore')
app_clicks['date'] = pd.to_datetime(app_clicks['date'], errors='ignore')

# ----------------------------------------------------
# 2. REVENUE + ORDERS BY COUNTRY
# ----------------------------------------------------
# Only completed orders count toward sales

orders_completed = order_header_df[order_header_df['status'] == 'COMPLETED']

country_sales = (
    orders_completed
    .groupby('country', as_index=False)
    .agg(
        revenue_eur=('amount_eur', 'sum'),
        orders=('order_id', 'nunique')
    )
)

# ----------------------------------------------------
# 3. CVR BY COUNTRY
# ----------------------------------------------------
# app_clicks: sid, country → gives us session-level country
# funnel_df: sessionid, converted → gives us funnel outcomes

# Get country per session (first occurrence)
session_country = (
    app_clicks[['sid', 'country']]
    .drop_duplicates(subset=['sid'])
    .rename(columns={'sid': 'sessionid'})
)

# Merge funnel results with country info
funnel_with_country = funnel_df.merge(session_country, on='sessionid', how='left')

# Compute CVR
cvr_country = (
    funnel_with_country
    .groupby('country', as_index=False)
    .agg(
        sessions=('sessionid', 'nunique'),
        converted_sessions=('converted', 'sum')
    )
)

cvr_country['cvr'] = cvr_country['converted_sessions'] / cvr_country['sessions']
cvr_country['cvr'] = cvr_country['cvr'].fillna(0)

# ----------------------------------------------------
# 4. MERGE ALL COUNTRY KPIs
# ----------------------------------------------------

country_kpis = (
    country_sales
    .merge(cvr_country[['country', 'cvr']], on='country', how='outer')
    .fillna(0)
)

# Sort by revenue for visualization
country_kpis = country_kpis.sort_values(by='revenue_eur', ascending=False)

print(country_kpis)

# ----------------------------------------------------
# 5. VISUALIZATION (Revenue + Orders + CVR)
# ----------------------------------------------------

fig, ax1 = plt.subplots(figsize=(12, 7))

x = country_kpis['country']

# Bar: Revenue (EUR)
ax1.bar(x, country_kpis['revenue_eur'], alpha=0.7, label='Revenue (EUR)')

# Bar: Orders (overlayed)
ax1.bar(x, country_kpis['orders'], alpha=0.7, label='Orders')

ax1.set_ylabel("Revenue (EUR) / Orders")
ax1.set_xlabel("Country")
ax1.tick_params(axis='x', rotation=45)

# Line: CVR
ax2 = ax1.twinx()
ax2.plot(x, country_kpis['cvr'], marker='o', linewidth=3, label='CVR', color='black')
ax2.set_ylabel("Conversion Rate (CVR)")

# Legends
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

plt.title("Sales by Area — Revenue, Orders & Conversion Rate (CVR) by Country")
plt.tight_layout()
plt.show()
```

# 2C) Seasonality

```python
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------
# 1. PREP: clean timestamps
# ----------------------------------------------------
df = order_header_df.copy()

# Convert to datetime (handles tz-aware and tz-naive)
df['createdate'] = pd.to_datetime(df['createdate'], errors='coerce')

# Drop invalid
df = df.dropna(subset=['createdate'])

# Keep only completed orders
df = df[df['status'] == 'COMPLETED'].copy()

# ----------------------------------------------------
# 2. Monthly aggregation
# ----------------------------------------------------
df['YearMonth'] = df['createdate'].dt.to_period('M')

monthly = (
    df.groupby('YearMonth')
      .agg(
          Monthly_Orders=('order_id', 'nunique'),
          Monthly_Revenue_EUR=('amount_eur', 'sum')
      )
      .reset_index()
)

monthly['YearMonth'] = monthly['YearMonth'].dt.to_timestamp()

# Rolling averages (3-month recommended)
monthly['Orders_Rolling'] = monthly['Monthly_Orders'].rolling(3, center=True).mean()
monthly['Revenue_Rolling'] = monthly['Monthly_Revenue_EUR'].rolling(3, center=True).mean()

print(monthly.head())
```

```python

# ----------------------------------------------------
# 3. PLOT — Monthly Revenue
# ----------------------------------------------------
plt.figure(figsize=(16,7))

plt.plot(monthly['YearMonth'], monthly['Monthly_Revenue_EUR'],
         label='Revenue (EUR)', color='green', alpha=0.3)

plt.plot(monthly['YearMonth'], monthly['Revenue_Rolling'],
         label='Revenue (3-mo rolling avg)', color='green', marker='o')

# Peak: Christmas (Dec)
decembers = monthly[monthly['YearMonth'].dt.month == 12]
if not decembers.empty:
    last_dec = decembers.iloc[-1]
    plt.annotate(
        "Christmas Peak",
        xy=(last_dec['YearMonth'], last_dec['Monthly_Revenue_EUR']),
        xytext=(last_dec['YearMonth'], last_dec['Monthly_Revenue_EUR']*1.15),
        arrowprops=dict(facecolor='black', arrowstyle='->'),
        ha='center'
    )

# Low season: Summer (Jul)
july = monthly[monthly['YearMonth'].dt.month == 7]
if not july.empty:
    last_jul = july.iloc[-1]
    plt.annotate(
        "Summer Low",
        xy=(last_jul['YearMonth'], last_jul['Monthly_Revenue_EUR']),
        xytext=(last_jul['YearMonth'], last_jul['Monthly_Revenue_EUR']*1.15),
        arrowprops=dict(facecolor='black', arrowstyle='->'),
        ha='center'
    )

plt.title("Monthly Revenue (EUR) — Seasonality with Rolling Average")
plt.xlabel("Month")
plt.ylabel("Revenue (EUR)")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()

```

```python

# ----------------------------------------------------
# 4. PLOT — Monthly Orders
# ----------------------------------------------------
plt.figure(figsize=(16,7))

plt.plot(monthly['YearMonth'], monthly['Monthly_Orders'],
         label='Orders', color='orange', alpha=0.3)

plt.plot(monthly['YearMonth'], monthly['Orders_Rolling'],
         label='Orders (3-mo rolling avg)', color='orange', marker='o')

# Peaks and lows again
if not decembers.empty:
    plt.annotate(
        "Christmas Peak",
        xy=(last_dec['YearMonth'], last_dec['Monthly_Orders']),
        xytext=(last_dec['YearMonth'], last_dec['Monthly_Orders']*1.15),
        arrowprops=dict(facecolor='black', arrowstyle='->'),
        ha='center'
    )

if not july.empty:
    plt.annotate(
        "Summer Low",
        xy=(last_jul['YearMonth'], last_jul['Monthly_Orders']),
        xytext=(last_jul['YearMonth'], last_jul['Monthly_Orders']*1.15),
        arrowprops=dict(facecolor='black', arrowstyle='->'),
        ha='center'
    )

plt.title("Monthly Orders — Seasonality with Rolling Average")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

# 2D) Sales Funnel (clicks → checkout intent → order)

```python
sessions['createdate'] = pd.to_datetime(sessions['createdate'], errors='coerce')
sessions.dropna(subset=['createdate'], inplace=True)
```

```python
funnel_counts = {
    'sessions': sessions['sessionid'].nunique(),
    'checkout_intent': sessions[sessions['processed'] == True]['sessionid'].nunique(),
    'orders': sessions[sessions['converted'] == True]['sessionid'].nunique()
}

funnel_rates = {
    'session_to_checkout': funnel_counts['checkout_intent'] / funnel_counts['sessions'],
    'checkout_to_order': funnel_counts['orders'] / funnel_counts['checkout_intent'],
    'total_cvr': funnel_counts['orders'] / funnel_counts['sessions']
}

print("FUNNEL COUNTS:", funnel_counts)
print("FUNNEL RATES:", funnel_rates)
```

```python
f_country = (
    sessions.groupby('country')
    .agg(
        sessions=('sessionid', 'nunique'),
        checkout=('processed', lambda x: (x==True).sum()),
        orders=('converted', lambda x: (x==True).sum())
    )
    .reset_index()
)

f_country['session_to_checkout'] = f_country['checkout'] / f_country['sessions']
f_country['checkout_to_order'] = f_country['orders'] / f_country['checkout'].replace(0, pd.NA)
f_country['total_cvr'] = f_country['orders'] / f_country['sessions']

print(f_country)
```

```python
f_language = (
    sessions.groupby('language')
    .agg(
        sessions=('sessionid', 'nunique'),
        checkout=('processed', lambda x: (x==True).sum()),
        orders=('converted', lambda x: (x==True).sum())
    )
    .reset_index()
)

f_language['session_to_checkout'] = f_language['checkout'] / f_language['sessions']
f_language['checkout_to_order'] = f_language['orders'] / f_language['checkout'].replace(0, pd.NA)
f_language['total_cvr'] = f_language['orders'] / f_language['sessions']

print(f_language)
```

```python
exit_points = {
    'dropped_before_checkout': len(sessions[sessions['processed'] == False]),
    'checkout_but_no_order': len(sessions[(sessions['processed'] == True) & (sessions['converted'] == False)]),
    'converted': len(sessions[sessions['converted'] == True])
}

print(exit_points)
```

```python
print("sessions.csv date range:", sessions['createdate'].min(), "→", sessions['createdate'].max())
print("total sessions:", sessions['sessionid'].nunique())
print("total converted:", sessions['converted'].sum())
```

```python
steps = ['Sessions', 'Checkout Intent', 'Orders']
values = [funnel_counts['sessions'], funnel_counts['checkout_intent'], funnel_counts['orders']]

plt.figure(figsize=(8,5))
plt.plot(steps, values, marker='o', linewidth=3)
plt.title('Sales Funnel (Sessions → Checkout → Orders)')
plt.ylabel('Sessions')
plt.grid(True)
plt.show()
```
